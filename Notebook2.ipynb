```python
import os
import pandas as pd
from transformers import pipeline

def main():
    # Using a specialized summarization model for better results
    # Switching to BART (which is BERT-like but for generation)
    summarizer = pipeline("text-generation", model="facebook/bart-large-cnn")

    files = ['Eng_train.csv', 'article_highlights.csv', 'Eng_validate.csv', 'Guj_train.csv', 'Guj_val_article.csv', 'Hin_train.csv', 'Hin_val_article.csv', 'axis_test.csv', 'axis_validation.csv', 'comparisons_train.csv', 'comparisons_validation.csv']

    for f_name in files:
        if os.path.exists(f_name):
            # Read CSV and target the text column
            df = pd.read_csv(f_name)
            # Find the column containing 'text' or 'article'
            text_col = [c for c in df.columns if 'text' in c.lower() or 'article' in c.lower()]

            for text in df[text_col].head(3): # Processing first 3 for space efficiency
                # 1. Run the model
                raw_output = summarizer(text, max_length=40, min_length=30, do_sample=False)

                # 2. Extract ONLY the summary text from the dictionary
                # The pipeline returns: [{'summary_text': 'The actual summary here...'}]
                summary_only = raw_output[0]

                # 3. Clean and Print
                print(f"SUMMARY: {summary_only['generated_text']}")
                print("-" * 30)

if __name__ == "__main__":
    main()
```
